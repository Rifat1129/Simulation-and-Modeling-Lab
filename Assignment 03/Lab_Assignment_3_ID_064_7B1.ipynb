A customer service center studies how long customers wait before being served.

They suspect:
1. On regular days, customers arrive randomly and wait times are uniformly distributed
between 2 and 12 minutes.
2. On busy days, wait times follow a normal distribution with an average (mean) of 8
minutes and a standard deviation of 2 minutes.

Your task is to simulate and analyze both situations using Python.


**Problem Statement :** Write a Python program to simulate and compare customer wait times for regular and busy days.

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
"""
Part 1 — Data Generation

  1. Generate 3,000 wait times using a uniform distribution between 2 and 12 minutes.
  2. Generate 3,000 wait times using a normal distribution with:
    o Mean = 8
    o Std Dev = 2
  3. Ensure no negative wait times in the normal distribution (replace negatives with 0).

"""

n = 3000

# Uniform Distribution (Regular Days): between 2 and 12 minutes
regular_days = np.random.uniform(2, 12, n)

# Normal Distribution (Busy Days): Mean = 8, Std Dev = 2
busy_days = np.random.normal(8, 2, n)

# Ensure no negative wait times (Replace negatives with 0)
busy_days = np.maximum(busy_days, 0)
```

```python
"""
Part 2 — Visualization

  1. Plot histograms for both distributions.
  2. Show them:
    o In the same figure (two subplots), OR
    o Overlaid with transparency.

"""

plt.figure(figsize=(12, 5))

# Subplot 1: Regular Days (Uniform)
plt.subplot(1, 2, 1)
sns.histplot(regular_days, bins=30, color='Green', kde=True)
plt.title('Wait Times - Regular Days (Uniform Distribution)')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')

# Subplot 2: Busy Days (Normal)
plt.subplot(1, 2, 2)
sns.histplot(busy_days, bins=30, color='Pink', kde=True)
plt.title('Wait Times - Busy Days (Normal Distribution)')
plt.xlabel('Wait Time (minutes)')
plt.ylabel('Frequency')

plt.tight_layout()
plt.show()
```

```python
"""
Part 3 — Statistical Analysis

For both datasets calculate:
  1. Mean
  2. Median
  3. Standard deviation
  4. Maximum wait time

"""

def calculate_stats(data, label):

    print(f"--- Statistics for {label} ---")
    print(f"Mean              : {np.mean(data):.4f}")
    print(f"Median            : {np.median(data):.4f}")
    print(f"Standard Deviation: {np.std(data):.4f}")
    print(f"Maximum Wait Time : {np.max(data):.4f}")
    print("------------------------------------------")
    print("\n")

calculate_stats(regular_days, "Regular Days (Uniform)")
calculate_stats(busy_days, "Busy Days (Normal)")
```
