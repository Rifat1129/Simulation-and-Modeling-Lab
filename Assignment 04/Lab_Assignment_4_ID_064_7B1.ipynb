```python
import random
import numpy as np
import math
```

```python
# Part 1 : Setup & Personalization.

R = 64
radius = R / 10
name = "Rifat Hassan"
letter = len(name.replace(" ", ""))
base_points = 200 * letter

print(f"R (last two digits)      : {R}")
print(f"Radius (R/10)            : {radius}")
print(f"Full name                : {name}")
print(f"Letters (no spaces)      : {letter}")
print(f"Base total points (N0)   : {base_points}")
```

```python
# Part 2 :  Simulation Logic

def is_inside_circle(x, y, r):
    return x * x + y * y <= r * r
```

```python
def run_simulation(total_points, R):
    r = R / 10
    low, high = 0.0, r

    inside = 0
    outside = 0

    half = total_points // 2

    for _ in range(half):
        x = random.uniform(low, high)
        y = random.uniform(low, high)
        if is_inside_circle(x, y, r):
            inside += 1
        else:
            outside += 1

    rem = total_points - half
    xs = np.random.rand(rem) * r
    ys = np.random.rand(rem) * r

    for x, y in zip(xs, ys):
        if is_inside_circle(x, y, r):
            inside += 1
        else:
            outside += 1

    pi_est = 4.0 * inside / total_points
    error = abs(pi_est - math.pi)

    return inside, outside, pi_est, error
```

```python
# Part 3 : Output and Analysis


runs = [base_points * k for k in range(1, 6)]

header = (
    f"{'Range Used':<15} | "
    f"{'Total Points':<12} | "
    f"{'Points Inside':<13} | "
    f"{'Estimated π':<12} | "
    f"{'Error from math.pi':<18}"
)
print(header)
print("-" * len(header))

for N in runs:
    inside, outside, pi_est, error = run_simulation(N, R)
    range_str = f"[0, {radius}]"
    print(
        f"{range_str:<15} | "
        f"{N:<12} | "
        f"{inside:<13} | "
        f"{pi_est:<12.8f} | "
        f"{error:<18.8f}"
    )
```
